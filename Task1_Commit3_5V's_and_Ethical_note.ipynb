```python
# ── Cell 1: SparkSession Initialisation (CW mandatory) ────────────────────────
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("7006SCN_HMDA2023_Task1_DataUnderstanding")
    .master("local[*]")
    .config("spark.driver.memory",          "10g")
    .config("spark.sql.shuffle.partitions", "200")
    .config("spark.driver.extraJavaOptions","-Xss4m")
    .getOrCreate()
)

spark.sparkContext.setLogLevel("WARN")

print(f"Spark {spark.version}  UI: {spark.sparkContext.uiWebUrl}")
```

```python
# ── Cell 2: Dataset Verification + File-Size Validation (CW mandatory) ─────────
# Commit 2: feat(task1): HMDA 2023 dataset verified — 11.56M rows, 99+ columns, 4.39GB
# Source : HMDA 2023 National Loan-Level Dataset (CFPB/FFIEC)
# URL    : https://ffiec.cfpb.gov/data-publication/snapshot-national-loan-level-dataset/2023

import subprocess, os

CSV_PATH = f"{DATA}/My_Data.csv"

if os.path.exists(CSV_PATH):
    print(f"Dataset located ✓ : {CSV_PATH}")
else:
    raise FileNotFoundError(f"Dataset not found: {CSV_PATH}")

print("\n=== FILE SIZE VERIFICATION — screenshot for Task 1 ===")
print(subprocess.run(["ls", "-lh", DATA], capture_output=True, text=True).stdout)

sz = os.path.getsize(CSV_PATH) / 1e9

lc = int(
    subprocess.run(
        ["wc", "-l", CSV_PATH],
        capture_output=True,
        text=True
    ).stdout.strip().split()[0]
) - 1

print(f"File size : {sz:.2f} GB")
print(f"Data rows : {lc:,}")

assert sz >= 1.0, "Dataset must be at least 1 GB"
assert lc >= 10_000_000, "Dataset must contain at least 10 million rows"

print("Big Data compliance: PASS ✓ (11,564,178 rows, 4.39 GB, Official Government Dataset)")
```

```python
# ── Cell 3: Schema + df.show(20) (CW mandatory screenshots) ───────────────────
# Commit 2: feat(task1): load HMDA dataset, inspect schema and preview records

from pyspark.sql import functions as F

df_raw = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(CSV_PATH)
    .cache()
)

print("="*70)
print("HMDA 2023 DATASET SUMMARY")
print("="*70)
print(f"Total rows    : {df_raw.count():,}")
print(f"Total columns : {len(df_raw.columns)}")
print("="*70)

print("=== df.printSchema() — screenshot for Task 1 ===")
df_raw.printSchema()
print()

print("=== df.show(20) — screenshot for Task 1 ===")
df_raw.show(20, truncate=True)

print("<-- Screenshot BOTH and embed in Task 1 report")
```

```python
# ── Cell 4: Label Engineering + 5 V's + Ethical Note ─────────────────────────
from pyspark.sql.functions import col, when

# ── Binary classification target construction ──────────────────────────────────
# loan_approval = 1 (APPROVED) if action_taken = Loan originated
# loan_approval = 0 (DENIED) if action_taken = Application denied
# All other application outcomes are excluded to create a clean binary
# classification dataset.

df_labeled = (
    df_raw
    .filter(col("action_taken").isin(1, 3))
    .withColumn(
        "loan_approval",
        when(col("action_taken") == 1, 1)
        .otherwise(0)
        .cast("double")
    )
)

print("=== TARGET: loan_approval distribution ===")
df_labeled.groupBy("loan_approval").count().orderBy("loan_approval").show()

print("""
=== 5 V's OF BIG DATA — HMDA 2023 National Loan-Level Dataset ===

Volume   : 11,564,178 mortgage loan application records (~4.39 GB CSV).
           Processing this dataset efficiently requires distributed computing
           using Apache Spark instead of in-memory tools such as pandas.

Velocity : HMDA data is collected annually from thousands of financial
           institutions across the United States. Although the public release
           is annual, lenders continuously generate loan application records
           during normal business operations.

Variety  : More than 99 structured attributes describing applicants, loans,
           properties and lenders, including categorical variables (loan type,
           occupancy type, applicant ethnicity), numerical variables (loan
           amount, income, interest rate), and geographic identifiers (state,
           county, census tract).

Veracity : Data is submitted by regulated financial institutions under the
           Home Mortgage Disclosure Act (HMDA). While subject to regulatory
           validation, some fields may contain missing values, reporting
           inconsistencies or values marked as Not Applicable.

Value    : Predicting mortgage approval decisions can support financial
           institutions by improving loan processing efficiency, identifying
           factors influencing approvals, supporting risk assessment and
           enhancing data-driven lending decisions.

=== ETHICAL & LICENSING NOTE ===
Licence   : Public dataset published by the Consumer Financial Protection Bureau
            (CFPB) and the Federal Financial Institutions Examination Council
            (FFIEC).

Source    : HMDA 2023 National Loan-Level Dataset.

Ethics    : The dataset contains sensitive financial and demographic
            information. Personally identifiable information is excluded from
            modelling, and results will only be reported in aggregate to avoid
            discrimination or privacy concerns.

NOT Kaggle: Downloaded directly from the official CFPB/FFIEC publication portal.

NOT teaching: This dataset has not been used in any 7006SCN lectures or
              laboratory sessions.
""")


print("Task 1 complete ✓")
```
